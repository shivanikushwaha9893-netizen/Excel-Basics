# Excel Basics


1. Explain the difference between Relative , Absolute and Mixed Cell Referencing.
   - In relative referencing, both row and column change; in absolute referencing, both row and column remain fixed; and in mixed referencing, either the row or the column is fixed while the other changes.


2. Write a formula to calculate the total sales of Car and Bicyle only only.
   - =SUMIF(A2:A4,"Car",B2:B4) + SUMIF(A2:A4,"Bicycle",B2:B4)


3. Using the data below, write a formula to calculate the average sales of items priced above 100 but less than 300.
   - =AVERAGEIFS(C2:C4, B2:B4, ">100", B2:B4, "<300")
